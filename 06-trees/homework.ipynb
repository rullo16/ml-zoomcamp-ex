```python
import pandas as pd
import numpy as np

import seaborn as sns
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
df = pd.read_csv('car_fuel_efficiency_2026.csv')
```

```python
df = df.fillna(0)
```

```python
from sklearn.model_selection import train_test_split
```

```python
df_full_train, df_test = train_test_split(df, test_size = 0.2, random_state = 1)
df_train, df_val = train_test_split(df_full_train, test_size = 0.2, random_state = 1)
```

```python
df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)
```

```python
df_train.columns
```

```python
df_train.dtypes
```

```python
y_train = df_train['fuel_efficiency_mpg']
y_val = df_val['fuel_efficiency_mpg']
y_test = df_test['fuel_efficiency_mpg']

del df_train['fuel_efficiency_mpg']
del df_val['fuel_efficiency_mpg']
del df_test['fuel_efficiency_mpg']

df_train = df_train.to_dict(orient='records')
df_val = df_val.to_dict(orient='records')
df_test = df_test.to_dict(orient='records')
```

```python
from sklearn.feature_extraction import DictVectorizer
dv=DictVectorizer(sparse=False)
```

```python
X_train = dv.fit_transform(df_train)
X_val = dv.transform(df_val)
X_test = dv.transform(df_test)
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
dtr = DecisionTreeRegressor(max_depth=1, random_state=1)
dtr.fit(X_train, y_train)
```

```python
from sklearn.tree import export_text
print(export_text(dtr, feature_names=list(dv.get_feature_names_out())))
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rf = RandomForestRegressor(n_estimators=10, random_state=1, n_jobs=-1)
rf.fit(X_train, y_train)
```

```python
from sklearn.metrics import root_mean_squared_error
```

```python
y_pred = rf.predict(X_val)
root_mean_squared_error(y_val, y_pred)
```

```python
errors = {}
for n_estimator in [10,50,100,150]:
    rf = RandomForestRegressor(n_estimators=n_estimator, random_state=1, n_jobs=-1)
    rf.fit(X_train, y_train)
    y_pred = rf.predict(X_val)
    errors[n_estimator] = root_mean_squared_error(y_val, y_pred)

from collections import defaultdict

by_depth = defaultdict(list)
for n_est, rmse in errors.items():
    by_depth[n_est].append(rmse)

for n_est in sorted(by_depth):
    mean_rmse = sum(by_depth[n_est]) / len(by_depth[n_est])
    print(f"n_estimators={n_est}: mean RMSE = {mean_rmse:.4f}")
```

```python
errors = {}
for m_d in [10,15,20,25]:
    for n_estimator in [10,50,100,150]:
        rf = RandomForestRegressor(n_estimators=n_estimator, random_state=1, n_jobs=-1, max_depth=m_d)
        rf.fit(X_train, y_train)
        y_pred = rf.predict(X_val)
        errors[(n_estimator, m_d)] = root_mean_squared_error(y_val, y_pred)

from collections import defaultdict

by_depth = defaultdict(list)
for (n_est, m_d), rmse in errors.items():
    by_depth[m_d].append(rmse)

for m_d in sorted(by_depth):
    mean_rmse = sum(by_depth[m_d]) / len(by_depth[m_d])
    print(f"max_depth={m_d}: mean RMSE = {mean_rmse:.4f}")
```

```python
rf = RandomForestRegressor(n_estimators=10, max_depth=20, random_state=1, n_jobs=-1)
rf.fit(X_train, y_train)

importances = pd.DataFrame({
    'feature': dv.get_feature_names_out(),
    'importance': rf.feature_importances_,
}).sort_values('importance', ascending=False)

print(importances)
```

```python
import xgboost as xgb
```

```python
dtrain = xgb.DMatrix(X_train, label=y_train, feature_names=list(dv.get_feature_names_out()))
dval = xgb.DMatrix(X_val, label=y_val, feature_names=list(dv.get_feature_names_out()))

watchlist = [(dtrain, 'train'), (dval, 'val')]

xgb_params = {
    'eta': 0.3,
    'max_depth': 6,
    'min_child_weight': 1,

    'objective': 'reg:squarederror',
    'nthread': 8,

    'seed': 1,
    'verbosity': 1,
}

eval_res = {}
model_03 = xgb.train(xgb_params, dtrain, num_boost_round=100, evals=watchlist, evals_result=eval_res, verbose_eval=False)
final_03 = eval_res['val']['rmse'][-1]

eval_res = {}
xgb_params['eta'] = 0.1
model_01 = xgb.train(xgb_params, dtrain, num_boost_round=100, evals=watchlist, evals_result=eval_res, verbose_eval=False)
final_01 = eval_res['val']['rmse'][-1]

print(f'eta=0.3 final val RMSE: {final_03:.4f}')
print(f'eta=0.1 final val RMSE: {final_01:.4f}')
print('Best:', 0.1 if final_01 < final_03 else 0.3)
```
