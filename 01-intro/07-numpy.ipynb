# Dot Product

```python
def vector_multiplication(u,v):
    assert u.shape[0] == v.shape[0]

    n = u.shape[0]
    res = 0.0
    for i in range(n):
        res += u[i]*v[i]
    return res
```

```python
u = np.array([2,4,5,6])
v = np.array([1,0,0,2])

vector_multiplication(u,v)
```

```python
u.dot(v)
```

# Matrix to Vector Mult

```python
def matrix_vector_mult(u,v):
    assert u.shape[1] == v.shape[0]
    n = u.shape[0]

    res = np.zeros(u.shape[0])
    for i in range(n):
        res[i] += vector_multiplication(u[i],v)
    return res
```

```python
u = np.array([[2,4,5,6],[1,2,1,2],[3,1,2,1]])
v = np.array([1,0.5,2,1])

matrix_vector_mult(u,v)
```

```python
u.dot(v)
```

# Matrix to Matrix Mult

```python
def matrix_mult(U,V):
    assert U.shape[1] == V.shape[0]

    rows = U.shape[0]
    cols = V.shape[1]

    res = np.zeros((rows,cols))

    for i in range(cols):
        vi = V[:,i]
        Uvi= matrix_vector_mult(U, vi)
        res[:,i] = Uvi
    return res

```

```python
U = np.array([[2,4,5,6],[1,2,1,2],[3,1,2,1]])
V = np.array([[1,1,2],[0,0.5,1],[0,2,1],[2,1,0]])

matrix_mult(U,V)
```

```python
U.dot(V)
```

# Inverse Matrix

```python
V = np.array([
    [1, 1, 2],
    [0, 0.5, 1],
    [0, 3, 1]
])
V_inv = np.linalg.inv(V)
V_inv
```
